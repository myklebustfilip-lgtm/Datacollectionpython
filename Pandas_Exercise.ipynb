Perform the following tasks:




1. Scrape the table titled: "List of countries and territories by total population" and create a DataFrame named **population_df**.

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
import matplotlib.pyplot as plt

# Step 1: Request the page content and parse with BeautifulSoup
url = "https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population"
data = requests.get(url).text
data
```

```python
soup = BeautifulSoup(data, 'html.parser')
soup
```

```python
#find all html tables in the web page
tables = soup.find_all('table')
tables# in html table is represented by the tag <table>
```

```python
len(tables)
```

```python
tables[0]
```

```python
tables[1]
```

```python
for index,table in enumerate(tables):
    if ("List of countries and territories by total population" in str(table)):
        table_index = index
print(table_index)
```

```python
print(tables[table_index].prettify())
```

```python
# Initialize an empty list to accumulate rows
rows_list = []

# Loop through the table rows and collect the data
for row in tables[table_index].tbody.find_all("tr"):
    col = row.find_all("td")
    if col:  # Check if the row contains any data
        rank = col[0].text.strip()
        location= col[1].text.strip()
        population = col[2].text.strip()
        per_of_world = col[3].text.strip()
        date = col[4].text.strip()

        # Add the row as a dictionary to the list
        rows_list.append({"Rank": rank, "Location": location, "Population": population, "% of world": per_of_world, "Date": date})

# Convert the list of dictionaries into a DataFrame
population_df = pd.DataFrame(rows_list)

# Display the resulting DataFrame
population_df
```

```python
# Easier way:
population_df = pd.read_html(url, match="List of countries and territories by total population", flavor='bs4')[0]
```

**API for Data Extraction**

```python
import openmeteo_requests

import requests_cache
import pandas as pd
from retry_requests import retry

# Setup the Open-Meteo API client with cache and retry on error
cache_session = requests_cache.CachedSession('.cache', expire_after = 3600)
retry_session = retry(cache_session, retries = 5, backoff_factor = 0.2)
openmeteo = openmeteo_requests.Client(session = retry_session)

# Make sure all required weather variables are listed here
# The order of variables in hourly or daily is important to assign them correctly below
url = "https://api.open-meteo.com/v1/forecast"
params = {
	"latitude": input("latitude:"),
	"longitude": input("longitude"),
	"hourly": "temperature_2m",
	"models": "metno_seamless"
}
responses = openmeteo.weather_api(url, params=params)

# Process first location. Add a for-loop for multiple locations or weather models
response = responses[0]
print(f"Coordinates {response.Latitude()}°N {response.Longitude()}°E")
print(f"Elevation {response.Elevation()} m asl")
#print(f"Timezone {response.Timezone()} {response.TimezoneAbbreviation()}")
#print(f"Timezone difference to GMT+0 {response.UtcOffsetSeconds()} s")

# Process hourly data. The order of variables needs to be the same as requested.
hourly = response.Hourly()
hourly_temperature_2m = hourly.Variables(0).ValuesAsNumpy()

hourly_data = {"date": pd.date_range(
	start = pd.to_datetime(hourly.Time(), unit = "s", utc = True),
	end = pd.to_datetime(hourly.TimeEnd(), unit = "s", utc = True),
	freq = pd.Timedelta(seconds = hourly.Interval()),
	inclusive = "left"
)}
hourly_data["temperature_2m"] = hourly_temperature_2m

hourly_dataframe = pd.DataFrame(data = hourly_data)
hourly_dataframe.head()

from matplotlib import pyplot as plt
import seaborn as sns
def _plot_series(series, series_name, series_index=0):
  palette = list(sns.palettes.mpl_palette('Dark2'))
  xs = series['date']
  ys = series['temperature_2m']

  plt.plot(xs, ys, label=series_name, color=palette[series_index % len(palette)])

fig, ax = plt.subplots(figsize=(10, 5.2), layout='constrained')
df_sorted = hourly_dataframe.sort_values('date', ascending=True)
_plot_series(df_sorted, '')
sns.despine(fig=fig, ax=ax)
plt.xlabel('date')
_ = plt.ylabel('temperature_2m')
```
