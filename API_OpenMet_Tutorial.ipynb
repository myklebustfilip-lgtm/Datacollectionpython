# 🌦️ Learning APIs: Fetching Weather Data with Open-Meteo

Welcome to this tutorial on using APIs! **API** stands for Application Programming Interface. It's how different software programs talk to each other. 

Today, we will learn how to use the **[Open-Meteo API](https://open-meteo.com/)**. Open-Meteo is a fantastic, free, open-source weather API that requires no API keys or registration for non-commercial use, making it perfect for learning.

### 🎯 What you will learn:
1. How to make an HTTP `GET` request in Python.
2. How to read and extract data from a JSON response.
3. How to organize API data using `pandas`.
4. How to visualize weather forecasts using `matplotlib`.

## Step 1: Import the necessary libraries
We need three standard libraries for this lesson:
- `requests`: To send HTTP requests to the Open-Meteo API.
- `pandas`: To organize the weather data into a clean table (DataFrame).
- `matplotlib`: To draw pretty graphs of our weather forecast.

```python
import requests
import pandas as pd
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## Step 2: Make Your First API Call (Current Weather)

To ask Open-Meteo for weather data, we need to provide the **latitude** and **longitude** of the location we want. 

Let's check the weather for **Oslo, Norway** (Latitude: `59.91`, Longitude: `10.75`). We will ask the API for the `current_weather`.

```python
# 1. Define the API endpoint URL
url = "https://api.open-meteo.com/v1/forecast"

# 2. Define the parameters (the questions we are asking the API)
params = {
    "latitude": 59.91,
    "longitude": 10.75,
    "current_weather": "true" # Ask for current weather conditions
}

# 3. Send the GET request
response = requests.get(url, params=params)

# 4. Check if the request was successful (Status Code 200 means OK!)
if response.status_code == 200:
    print("Success! Data retrieved.")
else:
    print(f"Error: {response.status_code}")
```

## Step 3: Understanding JSON Data
APIs usually return data in a format called **JSON** (JavaScript Object Notation). In Python, JSON looks and behaves exactly like a standard Python **dictionary**.

```python
# Convert the response to JSON / Python Dictionary
data = response.json()

# Let's print the raw data to see what it looks like
print("Raw JSON response:")
print(data)

print("\n--- Extracted Information ---")
# Extract specific pieces of data using dictionary keys
current_temp = data['current_weather']['temperature']
wind_speed = data['current_weather']['windspeed']

print(f"Current Temperature in Oslo: {current_temp}°C")
print(f"Current Wind Speed in Oslo: {wind_speed} km/h")
```

## Step 4: Getting an Hourly Forecast
Now let's do something more advanced. Instead of just the current weather, let's ask for the **hourly temperature and precipitation** for the next few days. We will also set our `timezone` so the times line up locally.

```python
forecast_params = {
    "latitude": 59.91,
    "longitude": 10.75,
    "hourly": ["temperature_2m", "precipitation"], # List of variables we want
    "timezone": "Europe/Oslo"
}

forecast_response = requests.get(url, params=forecast_params)
forecast_data = forecast_response.json()

# Extract the hourly data dictionary
hourly_data = forecast_data['hourly']

# Let's see how many hours of data we received
num_hours = len(hourly_data['time'])
print(f"Received {num_hours} hours of forecast data!")
```

## Step 5: Organizing Data with Pandas
Looking at raw lists of hundreds of temperatures is hard for humans. Let's put this data into a **Pandas DataFrame** (which is essentially a clean spreadsheet built in Python).

```python
# Create a DataFrame directly from our hourly data dictionary
df = pd.DataFrame(hourly_data)

# Convert the 'time' column from strings to actual Date/Time objects
df['time'] = pd.to_datetime(df['time'])

# Show the first 5 rows of our new table
df.head()
```

## Step 6: Visualizing the Data
Tables are great, but charts are better! Let's use `matplotlib` to plot the temperature for the upcoming week.

```python
plt.figure(figsize=(12, 5))

# Plot time on the X-axis and temperature on the Y-axis
plt.plot(df['time'], df['temperature_2m'], color='orange', linewidth=2)

# Add labels and a title
plt.title("7-Day Hourly Temperature Forecast for Oslo", fontsize=16)
plt.xlabel("Date & Time", fontsize=12)
plt.ylabel("Temperature (°C)", fontsize=12)

# Add a grid to make it easier to read
plt.grid(True, linestyle='--', alpha=0.6)

# Display the plot
plt.tight_layout()
plt.show()
```
