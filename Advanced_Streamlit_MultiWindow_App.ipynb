## 📄 Create Advanced Streamlit App

```python

import streamlit as st
import pandas as pd
import numpy as np
import plotly.express as px

# Sidebar navigation
st.sidebar.title("Navigation")
page = st.sidebar.radio("Go to", ["Dashboard", "Data Explorer", "Upload & Visualize", "About"])

# Dummy data
df = pd.DataFrame({
    "Region": np.random.choice(["North", "South", "East", "West"], size=100),
    "Product": np.random.choice(["A", "B", "C"], size=100),
    "Sales": np.random.randint(100, 1000, size=100),
    "Date": pd.date_range(start="2023-01-01", periods=100)
})

if page == "Dashboard":
    st.title("📊 Sales Dashboard")
    st.metric("Total Sales", f"${df['Sales'].sum():,.0f}")
    st.bar_chart(df.groupby("Region")["Sales"].sum())

elif page == "Data Explorer":
    st.title("🔍 Data Explorer")
    region = st.selectbox("Select Region", df["Region"].unique())
    product = st.multiselect("Select Product", df["Product"].unique(), default=list(df["Product"].unique()))
    filtered = df[(df["Region"] == region) & (df["Product"].isin(product))]
    st.write(f"Showing {len(filtered)} records")
    st.dataframe(filtered)

elif page == "Upload & Visualize":
    st.title("📁 Upload Your Data")
    uploaded_file = st.file_uploader("Choose a CSV file", type="csv")
    if uploaded_file is not None:
        user_df = pd.read_csv(uploaded_file)
        st.success("File uploaded successfully!")
        st.write(user_df.head())
        if user_df.select_dtypes(include=[np.number]).shape[1] >= 2:
            x_axis = st.selectbox("X Axis", user_df.select_dtypes(include=[np.number]).columns)
            y_axis = st.selectbox("Y Axis", user_df.select_dtypes(include=[np.number]).columns)
            fig = px.scatter(user_df, x=x_axis, y=y_axis, title="Scatter Plot")
            st.plotly_chart(fig)

elif page == "About":
    st.title("📘 About This App")
    st.write("""
    This multi-page Streamlit app demonstrates:
    - Sidebar navigation
    - Metrics and charts
    - Filterable data explorer
    - CSV upload and plot
    - Custom visualizations
    """)
```
